# Lesson 24: Evaluating binary classification models

A logistic-regression model produces probabilities. After choosing a threshold. it produces predicted classes

To evaluate those classes, we use a confusion matrix


## Confusion matrix

Suppose we have 1,000 transaction:

|  | **Actually fraud** | **Actually non-fraud |
| :--- | :--- | :--- |
| Predicted fraud | $TP=80$ | $FP=90$ |
| Predicted non-fraud | $FN=20$ | $TN=810$ |

The terms mean:

  - **TP:** fraud correctly detected
  - **TN:** normal transaction correctly accepted
  - **FP:** normal transaction incorrectly blocked
  - **FN:** fraud incorrectly accepted


## Accuracy

$$ Accuracy = \frac{TP + TN}{TP + TN + FP + FN} $$

Therefore:

$$ Accuracy = \frac{80 + 810}{1000} = 0.89 $$

The accuracy is 89%

However, a model that predicts every transaction as non-fraud would have:

$$ Accuracy = \frac{900}{1000} = 90\% $$

It wuold have higher accuracy while detecting no fraud. This is why accuracy can be misleading for imbalanced data



## Precision

$$ Precision = \frac{TP}{TP + FP} $$

$$ Precision = \frac{80}{80 + 90} \approx 0.4706 $$

Interpretation:

`Among transactions classified as fraud, approximately 47.06\% are actually fraudulent`



## Recall

Recall is also called sensetivity or the true-positive rate:

$$ Recall = \frac{TP}{TP + FN} $$

$$ Recall = \frac{80}{80 + 20} = 0.80 $$

Interpretation:

`The model detects 80% of all fraudulent transactions`



## Specificity

$$ Specificity = \frac{TN}{TN + FP} $$

$$ Specificity = \frac{810}{810 + 90} = 0.90 $$

The model correctly accepts 90\% of legitimate transactions



## False-positive rate

$$ FPR = \frac{FP}{FP + TN} = 1 - Specificity $$

$$ FPR = \frac{90}{900} = 0.10 $$

Therefore, 10\% of legitimate transactions are incorrectly flagged



## F1-score

The F1-score balances precision and recall:

$$ F1 = 2\frac{Precision \cdot Recall}{Precision + Recall} $$

It can also be calculated as:

$$ F1 = \frac{2TP}{2TP + FP + FN} $$

For the example:

$$ F1 = \frac{160}{160 + 90 +20} = \frac{160}{270} \approx 0.5926 $$



## Effect of changing the threshold

Lowering the threshold usually causes:
  - *TP* to increase
  - *FP* to increase
  - *FN* to decrease
  - *TN* to decrease
  - recall to increase
  - specificity to decrease

Precision may increase or decrease depending on the data, but it commonly decreases when many additional false positives appear



## Business costs

Suppose:

  - one missed fraud costs 500 units
  - one false alert costs 10 units

For out model:

$$ Cost = FN \cdot 500 + FP \cdot 10 $$

$$ Cost = 20 \cdot 500 + 90 \cdot 10 = 10\ 900 $$

Consider a lower threshold with:

$$ TP = 90, \ FN = 10, \ FP = 180, \ TN = 720 $$

Its cost is:

$$ 10 \cdot 500 + 180 \cdot 10 = 6\ 800 $$

Its accuracy is lower:

$$ \frac{90 + 720}{1000} = 81\% $$

but its business cost is also lower. Therefore, the best threshold is not necessarily the threshold with the highest accuracy


## ROC curve and ROC-AUC

The ROC curve compares:

$$ TPR = Recall $$

with:

$$ FPR = 1 - Specificity $$

at different threshold

ROC-AUC summarizes the model's ranking ability:

  - $AUC = 1$: perfect ranking
  - $AUC = 0.5$: approximately random ranking
  - higher AUC generally means better separation of positive and negative observations

ROC-AUC evaluates probabilities across threshold, not one particular threshold



## Precision-recall curve

The precision-recall curve compares precision and recall at different thresholds

For rare events such as fraud, PR-AUC is often more informative than ROC-AUC because it focuseson the positive class and reveals the effect of false positives more clearly

The baseline PR-AUC is approximately equal to the positive-class prevalence. If 1% of transactions are fraudulent, a random model has a PR-AUC of approximately 0.01
