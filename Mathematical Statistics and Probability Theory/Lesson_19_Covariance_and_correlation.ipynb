# Lesson 19: Covariance and correlation

Chi-square examines relationship between categorial variables. For two numerical variables, we often begin with covariance and correlation

Examples include:
  * study time and examination score
  * transaction amount and risk score
  * income and loan size
  * model confidence and prediction error

## Covariance

Sample covariance is:

$$ Cov(X, Y) = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{n - 1} $$

It describes whether two variables tend to move together

  * Positive covariance: larger $X$ tends to appear with larger $Y$
  * Negative covariance: larger $X$ tends to appear with smaller $Y$
  * Covariance near zero: no clear linear movement together


## Pearson correlation

Pearson correlation standardizes covariance:

$$ r = \frac{Cov(X, Y)}{s_{X}s_{Y}} $$

It always lies between -1 and 1

$$ -1 \leq r \leq 1 $$

| **Value** | **Interpretation**                  |
| ---       | ---                                 |
| r ≈ 1     | Strong positive linear relationship |
| r ≈ -1    | Strong negative linear relationship |
| r ≈ 0     | Weak or no linear relationship      |


The exact interpretation of "weak" and "strong" depends on the subject and context


### Example

Suppose:

$$ X = [1, 2, 3, 4, 5] $$

$$ Y = [2, 4, 5, 4, 5] $$

The means are:

$$ \bar{x} = 3, \quad \bar{y} = 4 $$

| $x_i$ | $x_i-\bar{x}$ | $y_i$ | $y_i-\bar{y}$ | Product |
| ---: | ---: | ---: | ---: | ---: |
| 1 | -2 | 2 | -2 | 4 |
| 2 | -1 | 3 | 0 | 0 |
| 3 | 0 | 5 | 1 | 0 |
| 4 | 1 | 4 | 0 | 0 |
| 5 | 2 | 5 | 1 | 2 |


Therefore:

$$ Cov(X, Y) = \frac{4 + 0 + 0 + 0 + 2}{5 -1} = 1.5 $$

The sample standard deviations are approximately:

$$ s_X \approx 1.581, \quad s_Y \approx 1.225 $$

Thus:

$$ r = \frac{1.5}{1.581 \cdot 1.225} \approx 0.775 $$

There is a reasonably strong positive linear relationship


## Testing coorelation

The popultaion correlation is represented by *ρ*

We can test:

$$ H_0 : \rho = 0, $$

$$ H_1 : \rho \neq 0 $$

The test statistic is:

$$ t = r\sqrt{\frac{n-2}{1-r^2}} $$

with

$$ df = n - 2 $$

For the example:

$$ t = 0.775\sqrt{\frac{5-2}{1-0.775^2}} \approx 2.12 $$

Here:

$$ df = 5 - 2 = 3 $$

Using:

$$ t_{critical} = 3.182 $$

for two-sided test with $\alpha$ = 0.05

$$ |2.12| < 3.182 $$

We fail to reject $H_0$

This illustates an important point: a correlation may appear strong, but a very small sample provides insufficient evidence that the population correlation is non-zero
