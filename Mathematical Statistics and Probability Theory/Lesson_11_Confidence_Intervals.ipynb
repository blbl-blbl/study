# Lesson 11. Confidence Intervals

A sample mean x̄ is a **point estimate** of the unknown population mean μ.

However, a single number doesn't show how uncertain our estimation is. A confidence interval (CI) gives a range of plausible for μ.

When the population standard deviation σ is known, the confidence interval is:

$$ \bar{X} \pm z_{\alpha/2} \frac{\sigma}{\sqrt{n}} $$

Here:


*   x̄ is the sample mean
*   σ / √n is the standard error
*   zα/2 is the critical value
*   the quantity after ± is the margin error

Common critical values

| Confidence interval | Critical value |
| :---: | :---: |
| 90% | 1.645 |
| 95% | 1.960 |
| 99% | 2.576 |




## Example

Suppose:

$$ \bar{X} = 52, \quad \sigma = 20, \quad n = 100 $$


The standard error is:

$$ \frac{20}{\sqrt{100}}  = 2 $$


For a 95% confidence interval:

$$ 52 \pm 1.96 \cdot 2 $$


The margin of error is:

$$ ME = 1.96 \cdot 2 = 3.92 $$


Therefore:

$$ \text{CI}_{95\%} = [52 - 3.92, 52 + 3.92] = [48.08, 55.92] $$


### Correct interpretation

**If we repeatedely took samples and constructed intervals using this method,
 aproximately 95% of those intervals would contain the true population mean**


For practical communication, we often say:

**We are 95% confident that the true population mean lies between 48.08 and
55.92**

## What changes the interval width?

THe margin error is:

$$ ME = z_{\alpha/2} \frac{\sigma}{\sqrt{n}} $$

Therefore:
*   higher confidence level → wider interval
*   higher variability σ → wider interval
*   larger sample n → narrower interval

To halve the interval width, we need four times as many observations


## Required sample size

If we want a maximum margin of error ME, we can calculate:

$$ n = \left( \frac{z_{\alpha/2} \sigma}{ME} \right)^2 $$


Because the sample size must be an integer, we always round up

For example, if:

$$ \sigma = 20, ME = 2, z = 1.96 $$

then:

$$ n = \left( \frac{1.96 \cdot 20}{2} \right)^2 = 384.16 $$

We need at least:

$$ n= 385 $$

## Unknown population standard deviation

In practice, σ is usually unknown. We replace it with the sample standard deviation **s** and use the Student's **t-distribution**:

$$ \bar{X} \pm t_{\alpha/2, n-1} \frac{s}{\sqrt{n}}  $$

The t-distribution has heaver tails than the normal distribution, porducing wider intervals, especially for small samples
