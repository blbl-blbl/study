# Lesson 16: Paired *t*-test


A paired *t*-test is used when each observation in one condition is naturally connected to an observation in the other condition

Example include:
  * the same customers and after a change
  * the same employees before and after training
  * two models evaluated on the same transactions
  * matched experimental sub


## Pairwise difference

Suppose we measure every object before and after an intervention. For each pair, calculate
$$ D_i = X_{after,i} - X_{before,i} $$

The direction must be chosen consistently. Here, a positive difference means that the value increased

We then calculate the mean difference:
$$ \bar{D} = \frac{D_1 + ... + D_n}{n} $$

The hypotheses are:
$$ H_0 : \mu_D = 0, $$
$$ H_1 : \mu_D \neq 0 $$

The paired *t*-test is therefore simply a **one-sample *t*-test apllied to the differences**


## Standard deviation of the differences

First calculate:
$$ s_D = \sqrt{\frac{\sum_{i=1}^n(D_i - \bar{D})^2}{n-1}} $$

The standard error is:
$$ SE(\bar{D}) = \frac{s_D}{\sqrt{n}} $$

The test statistic is:
$$ t = \frac{\bar{D} - 0}{s_D/\sqrt{n}} $$

The degrees of freedom are:
$$ df = n-1 $$


## Example

Suppose the differences after minus before are:
$$ 2, 4, 1, 3, 5 $$

The mean difference is:
$$ \bar{D} = \frac{2 + 4 + 1 + 3 + 5}{5} = 3 $$

The deviations from the maen are:
$$ -1, 1, -2, 0, 2 $$

Therefore:
$$ s_D = \sqrt{\frac{(-1)^2+1^2+(-2)^2+0^2+2^2}{5-1}} = \sqrt{\frac{10}{4}} \approx 1.581 $$

The standard error is:
$$ SE = \frac{1.581}{\sqrt{5}} \approx 0.707 $$

The test statistic is:
$$ t = \frac{3}{0.707} \approx 4.243 $$

For *df* = 4, suppose:
$$ t_{critical} = \pm 2.776 $$

Because:
$$ |4.243| > 2.776 $$

we reject H_0

The confidence interval is:
$$ 3 \pm 2.776 \cdot 0.707 \approx [1.04, 4.96] $$

We estimate that the mean increase lies between approximately 1.04 and 4.96


## Why pairing can increase power

Different people or objects often hve naturally different baseline values. By calculating pairwise differences, we remove much of this between-object variation

However, pairing helps only when the observations are genuinely matched. We cannot arbitrarily pair unrelated observations.


## Assumptions

The paired *t*-test assumes:
  * the pairs are independent of other pairs
  * the outcome is quantitative
  * the pairwise differences are approximately normally distributed
  * the differences do not contain outliers

The normally assumption concerns the **differences**, not neccessarily the original before and after measurements

### Task 1

Six employees have the following productivity scores:

| Employee | Before | After |
| ---      | ---    | ---   |
| 1        | 70     | 74    |
| 2        | 72     | 73    |
| 3        | 68     | 72    |
| 4        | 75     | 78    |
| 5        | 65     | 69    |
| 6        | 80     | 82    |

Define:
$$ D_i = X_{after,i} - X_{before,i} $$

Test:
$$ H_0 : \mu_D = 0, H_1 : \mu_D \neq 0 $$

Use:
$$ t_{critical} = \pm 2.571 $$

Calculate:
  1. all six differences
  2. the mean difference
  3. the sample standard deviation of the differences
  4. the standard error
  5. the *t*-statistic
  6. the statistic decision
  7. the 95% confidence interval for $\mu_D$


**Solution:**
1.
| Employee | Before | After | Difference |
| ---      | ---    | ---   | ---        |
| 1        | 70     | 74    | 4          |
| 2        | 72     | 73    | 1          |
| 3        | 68     | 72    | 4          |
| 4        | 75     | 78    | 3          |
| 5        | 65     | 69    | 4          |
| 6        | 80     | 82    | 2          |

2. the mean difference:
$$ Mean~Diff = \frac{4+1+4+3+4+2}{6} = 3 $$

3. the sample standard deviation of the difference
$$ s_D = \sqrt{\frac{8}{6-1}} \approx 1.265 $$

4. the standard error:
$$ SE = \frac{1.265}{\sqrt{6}} \approx 0.516  $$

5. the *t*-statistic
$$ t = \frac{3}{0.516} \approx 5.81 $$

6. the statistic decision

Because

$$ |5.81| > 2.571 $$

we reject $H_0$

7. the 95% confidence interval for $\mu_D$
$$ CI_{95\%} = 3 \pm 2.571 \cdot 0.516 \approx [1.67, 4.33] $$
