# Lesson 21: Statistical inference for a regression slope

The estimated slope $b_1$ describes out sample. Now we want to determine whether the population slop $\beta_1$ is different from zero

## Hypothesis

For a two-sided test:

$$ H_0: \beta_1 = 0, $$

$$ H_1: \beta_1 \neq 0 $$

If:

$$ \beta_1 = 0, $$

there is no population linear relationship between *X* and *Y*


## Residual degrees of freedom

Simple linear regression estimates two parameters:

$$ b_0~and~b_1 $$

Therefore, the residual degrees of freedom are:

$$ df = n - 2 $$


## Mean squared error

The residual variance is estimated using:

$$ MSE = \frac{SSE}{n - 2} $$

The residual standard error is:

$$ s_e = \sqrt{MSE} = \sqrt{\frac{SSE}{n-2}} $$

It measures the typical prediction error in the units of *Y*


## Standard error of the slope

Define:

$$ S_{xx} = \sum(x_i - \bar{x})^2 $$

The standard error ob $b_1$ is:

$$ SE(b_1) = \frac{s_e}{\sqrt{S_{xx}}} $$

A smaller standard error means that the slope has been estimated more precisely


## Test statistic

The test statistic is:

$$ t = \frac{b_1 - 0}{SE(b_1)} $$

It follows a *t*-distibution with:

$$ df = n- 2 $$

### Example

For the previous regression:

$$ n = 5, \quad b_1 = 1.3 $$
$$ SSE = 1.1, \quad S_{xx} = 10 $$

First, calculate the residual degrees of freedom:

$$ df = 5 - 2 = 3 $$

Then:

$$ MSE = \frac{1.1}{3} \approx 0.3667 $$

The residual standard error:

$$ s_e = \sqrt{0.3667} \approx 0.6055 $$

The slope standard error:

$$ SE(b_1) = \frac{0.6055}{\sqrt{10}} \approx 0.1915 $$

Therefore:

$$ t = \frac{1.3}{0.1915} \approx 6.79 $$

For a two-sided test with $\alpha=0.05$ and $df=3$:

$$ t_{critical} = 3.182 $$

Since:

$$ |6.79| > 3.182 $$

we reject $H_0$

There is statistically significant evidence that the population slope is not zero


## Confidence interval for the slope

A confidence interval is:

$$ b_1 \pm t_{\alpha/2,n-2}SE(b_1) $$

For the example:

$$ 1.3 \pm 3.182(0.1915) = 1.3 \pm 0.609 $$

Therefore:

$$ CI_{95\%} \approx [0.691, 1.909] $$

Interpretation:

`We are 95% confident that a one-unit increase in X is associated with and average increase in Y of between approximately 0.691 and 1.909 units`

The interval does not contain zero, which agrees with the hypothesis-test decision


## Relationship with the correlation test

In simple linear regression:

$$ H_0 : \beta_1 = 0 $$

is equivalent to:

$$ H_1 : \rho = 0 $$

Therefore, the correlation test and slope test produce the same *t*-statistic and *p*-value


## Statistical versus practical significance

A slope may be statistically significant but practicully unimportant

For example:

$$ b_1 = 0.001 $$

may become statistically significant with millions of observations. We must interpret:

  * the slope's size and units
  * its confidence interval
  * statistical significance
  * practical of business importance
