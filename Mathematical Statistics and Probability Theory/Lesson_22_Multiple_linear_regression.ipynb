# Lesson 22: Multiple linear regression

Simple linear regression uses one predictor:

$$ Y = \beta_0 + \beta_1X + \epsilon $$

Mulitiple linear regression uses or more predictors:

$$ Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + ... + \beta_kX_k + \epsilon $$

The fitted model is:

$$ \hat{Y} = b_0 + b_1X_1 + b_2X_2 + ... + b_kX_k $$


## Coefficient interpretation

The most important phrase in multiple regression is:

| Holding all other predictors constant

Suppose a fraud-risk model is:

$$ \widehat{Risk} = 20 + 0.8Amount - 1.5Age $$

where:
  * *Amount* is the transaction amount in thousands
  * *Age* is the account age in years

The coefficient 0.8 means:

`Holding account age constant, a one-thousand-unit increase in transaction amount is associated with an average increase of 0.8 points in predicted risk`


## Prediction

For:

$$ Amount = 50, \quad Age = 4  $$

the predicted risk is:

$$ \widehat{Risk} = 20 + 0.8(50) - 1.5(4) = 20 + 40 - 6 = 54 $$


## Why coefficients may change

Suppose we first regress risk only on transaction amount:

$$ \widehat{Risk} = b_0 + b_1Amount $$

The amount coefficient may partly capture the effect of account age if newer accounts also tend to make larger transactions

After adding account age:

$$ \widehat{Risk} = b_0 + b_1Amount + b_2Age $$

the coefficient of amount describes its relationship with risk after controlling for age

Therefore, coefficients can change when new predictirs are added


## Categorical predictors

Categorical variables can be represented using indicator variables

Suppose:

$$ DefaultHustory = \begin{cases}1,&\text{previous default}, \\ 0,&\text{no previous default}\end{cases} $$

Consider:

$$ \widehat{Score} = 600 + 5Income - 40DefaultHistory $$

For customers with the same income, a previos default is associated with a predicted score that is 40 points lower

The category coded as zero is called the **reference category**


## Residual degrees of freedom

If the model contains *k* predictors and an intercept, it estimates:

$$ k + 1 $$

coefficients

Therefore:

$$ df_{residual} = n - k - 1 $$

For $n = 100$ and two predictors:

$$ df = 100 - 2 - 1 = 97 $$


## Testing individual coefficients

For each predictor:

$$ H_0 : \beta_{j} = 0 $$

$$ H_1 : \beta_{j} \neq 0 $$

The test statistic is:

$$ t = \frac{b_j}{SE(b_j)} $$

This test asks whether a predictor is associated with Y after controlling for all other predictors in the model


## $R^2$ and adjusted $R^2$

Regular $R^2$ is:

$$ R^2 = 1 - \frac{SSE}{SST} $$

Adding another predictor can never decrease $R^2$. even if the predictor is useless

Adjusted $R^2$ penalizes unnecessary predictors:

$$ R^2_{adj} = 1 - (1 - R^2)\frac{n-1}{n-k-1} $$

Unlike regular $R^2$, adjusted $R^2$ can decrease when a useless predictor is added


### Example

Suppose:

$$ n = 100, \quad k = 2, \quad R^2 = 0.64 $$

Then:

$$ R^2_{adj} = 1 - (1 - 0.64)\frac{99}{97} \approx 0.6326 $$

Thus, approximately 63.26% of the variation is explained after adjusting for the number of predictors


## Multicollinearity

Multicollinearity occurs when predictors are strongly correlated with each other

For example:

  * annual income
  * monthly income

These variables contain almost the same information

Multicollinearity can cause:

  * unstabe coefficients
  * large coefficients standard errors
  * unexpected coefficient signs
  * coefficients that change strongly when predictors are added or removed

The model may still predict well, but interpreting individual coefficients becomes difficult


## Assumptions

Multiple linear regression assumes:

  * a linear relationship between predictors and the expected value of *Y*
  * independent observations
  * approximately constant residual variance
  * no extreme influental observations
  * no perfect multicollinearity
  * approximately normal residuals for hypothesis tests and confidence intervals
