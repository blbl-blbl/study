# Lesson 17: Comparing two independent proportions


Previously, we compared **means** using *t*-tests. Now we will compare proportions, such as :
  * conversion rates in an A/B test
  * fraud-detection rates of two systems
  * default rates for two groups
  * errors rates of two models

The outcome must be binary: success/failture, fraud/not fraud, converted/not converted, and so on


## Sample proportions


Suppose:
  * Group A has $x_A$ successes among $n_A$ observations
  * Group B has $x_B$ successes among $n_B$ observations

Their sample proprotions are:
$$\hat{p}_A = \frac{x_A}{n_A}, \quad \hat{p}_B = \frac{x_B}{n_B} $$

The observed difference is:
$$ \hat{p}_B - \hat{p}_A $$

The order matters. If we calculate B - A, a positive result means that Group B has the higher rate


## Two-proportion $z$-test


For a two-sided test:
$$ H_0 : p_B - p_A = 0, $$
$$ H_1 : p_B - p_A \neq 0 $$

The null hypothesis says that the population proportions are equal


### Step 1: Calculate the pooled proportion

Under $H_0$, we assume that both groups have one common population proportion. Therefore, we combine their results:

$$ \hat{p}_{pool} = \frac{x_A + x_B}{n_A + n_B} $$


### Step 2: Calculate the standard error under $H_0$

$$ SE_{H_0} = \sqrt{\hat{p}_{pool}(1 - \hat{p}_{pool}) \left( \frac{1}{n_A} + \frac{1}{n_B} \right)} $$


### Step 3: Calculate the test statistic

$$ z = \frac{\hat{p}_B - \hat{p}_A}{SE_{H_0}} $$

For a two-sided test with α = 0.05

$$ z_{critical} = \pm 1.96 $$

Therefore:
  * if $|z| > 1.96$, reject $H_0$
  * if $|z| \leq 1.96$ fail to reject $H_0$


## Example

Suppose:
$$ x_A = 100, \quad n_A = 1000 $$
$$ x_B = 130, \quad n_B = 1000 $$

The sample proportions are:

$$ \hat{p}_A = \frac{100}{1000} = 0.10 $$
$$ \hat{b}_B = \frac{130}{1000} = 0.13 $$

The observed difference is:

$$ \hat{p}_B - \hat{p}_A = 0.13 - 0.10 = 0.03 $$

This is an increase of 3 **percentage points**

The pooled proportion is:

$$ \hat{p}_{pool} = \frac{100 + 130}{1000 + 1000} = \frac{230}{2000} = 0.115 $$

The standard error is:

$$ SE_{H_0} = \sqrt{0.115(1-0.115) \left( \frac{1}{1000} + \frac{1}{1000} \right)} \approx 0.01427 $$

Therefore:

$$ z = \frac{0.13 - 0.10}{0.01427} \approx 2.10 $$

Since:

$$ |2.10| > 1.96 $$

we reject $H_0$ at the 5% signnificance level. There is statistically significant evidence that the two population proportions are different


## Confidence interval for the difference

For a confidence interval, we do **not** pool the proportions:

$$ SE_{CI} = \sqrt{\frac{\hat{p}_A(1 - \hat{p}_A)}{n_A} + \frac{\hat{p}_B(1-\hat{p}_B)}{n_B}} $$

The 95% confidence interval is:

$$ (\hat{p}_B - \hat{p}_A) \pm 1.96SE_{CI} $$

For our example:

$$ SE_{CI} = \sqrt{\frac{0.10(0.90)}{1000} + \frac{0.13(0.87)}{1000}} \approx 0.01425 $$

Thus:

$$ 0.03 \pm 1.96(0.01425) = 0.03 \pm 0.02793 $$

$$ CI_{95\%} \approx [0.0021, 0.0579] $$

The interval does not contain zero, which agrees with the hypothesis-test result

We estimate that Group B's population rate is between approximately **0.21 and 5.79 percentage points higther** than Group A's rate.


## Pooled versus upooled standard errors

| Calculation         | Standard error |
| ---                 | ---            |
| Hypothesis test     | Pooled         |
| Confidence interval | Unpooled       |

We pool in the hypothesis test because $H_0$ assumes:

$$ p_A = p_B $$

We do not pool for the confidence interval because we are estimating the difference without assuming that the proportions are equal


## Percentage points versus relative increase

An increase from 10% to 13% is:

$$ 13% - 10% = 3 $$

percentage points

The ralative increase is:

$$ \frac{13% - 10%}{10%} = 0.30 = 30% $$

Therefore:

> The rate increased by 3 percentage points, or by 30% relative to its original value

These are different statements


## Assumptions

The two-propotion $z$-test requires:
  * a binary outcome
  * two independent groups
  * independent observations within each group
  * sufficiently mant success and failures

A common rule is that each group should contain at least 10 success and 10 failures

### Task 1: A/B conversion test

A company tests two website designs:
$$ x_A = 80, \quad n_A = 800, $$
$$ x_B = 104, \quad n_B = 800 $$

Using a two-sided significance level of α = 0.05
  1. Calculate $\hat{p}_A$ and $\hat{p}_B$
  2. Calculate $\hat{p}_B - \hat{p}_A$
  3. State $H_0$ and $H_1$
  4. Calculate the pooled proportion
  5. Calculate $SE_{H_0}$
  6. Calculate the $z$-statistic
  7. Compare it with ±1.96 and make a dicision
  8. Construct the 95% confidence interval using the **unpooled** standard error
  9. Explain whether the statistical result proves that Design B is practically important


**Solution:**

1. Sample proportions by groups:

$$ \hat{p}_A = \frac{80}{800} = 0.1 $$
$$ \hat{p}_B = \frac{104}{800} = 0.13 $$

2. The observed difference is:

$$ \hat{p}_B - \hat{p}_A = 0.13 - 0.10 = 0.03 $$

3. For two-sided test:

$$ H_0 : \hat{p}_B - \hat{p}_A = 0 $$
$$ H_1 : \hat{p}_B - \hat{p}_A \neq 0 $$

4. The pooled proportion:

$$ \hat{p}_{pool} = \frac{80 + 104}{800 + 800} = 0.115 $$

5. The standard error under $H_0$

$$ SE_{H_0} = \sqrt{0.115(1-0.115) \left( \frac{1}{800} + \frac{1}{800} \right) } = $$

$$ = \sqrt{0.101775 \cdot 0.0025} \approx 0.01595 $$

6. The $z$-statistic

$$ z = \frac{0.13 - 0.10}{0.01595} \approx 1.88 $$

7. Because:
$$ |1.88| < 1.96 $$

we fail to reject $H_0$ at the 5% significance level

8. The 95% confidence interval

$$ SE_{CI} = \sqrt{\frac{0.1(1-0.1)}{800} + \frac{0.13(1-0.13)}{800}} = $$

$$ = \sqrt{0.0001125 + 0.000141375} \approx 0.01593 $$


$$ CI_{95\%} = 0.03 \pm 1.96 \cdot 0.01593 \approx [-0.0012, 0.0612] $$


9.
The interval contain zero, which agrees with the hypothesis-test result

### Task 2: One-sided fraud-detection test

An old fraud system detects:

$$ 180~of~1000~fraudulent~transactions $$

A new independent system detects:
$$ 215~of~1000~fraudulent~transactions $$

We want to determine whether the new system has a highter detectons rate

  1. Calculate both sample proportions
  2. State the hypotheses:
    $$ H_0 : p_{new} - p_{old} = 0 $$
  and formulate the appropriate one-sided alternative
  3. Calculate the pooled proportion
  4. Calculate $SE_{H_0}$
  5. Calculate the $z$-statistic
  6. Use $z_{critical}=1.645$ for α = 0.05 and make a decision
  7. Explain the result in one sentence


**Solution:**

1. $$ p_{old} = \frac{180}{1000} = 0.18, \quad p_{new} = \frac{215}{1000} = 0.215 $$

2. $$ H_1 : p_{new} - p_{old} > 0 $$

3. $$ p_{pool} = \frac{180 + 215}{1000 + 1000} = 0.1975 $$

4. $$ SE_{H_0} = \sqrt{0.1975(1-0.1975) \left( \frac{1}{1000} + \frac{1}{1000} \right)} = \sqrt{0.15849375 \cdot 0.002} \approx 0.0178 $$


5. $$ z = \frac{0.215-0.18}{0.0178} \approx 1.966 $$

6. Because:

$$ |1.966| > 1.645 $$

we reject $H_0$

7. There is statistically significant evidence that the new system has a higher fraud-detection rate than the old system.

### Task 3: Conceptual questions

  1. Why do we use a pooled proportion for the hypothesis test? - because in the hypothesis $H_0 : p_A = P_B$, it allows us to find weighted standard error of samples, and then we can determine *z*-statistic for comparing it

  2. Why do we use an upooled standard error for the confidence interval? - because we estimating the difference between proportions

  3. What the difference between a 3-percentage-point and a 3% relative increase? - this is different formulas with different purposes: increase: $10\% - 7\% = 3\%$, relative increase: $(10\% - 7\%) / 7\% = 42.86% $

  4. Can we use this test if the same customers receive both Design A and Design B? - No, we can't, groups must be independent

  5. What problem can occur when there are very few successes? - the problem is that the normal approximation used by the z-test becomes unreliable.
