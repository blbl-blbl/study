# Lesson 17: Comparing two independent proportions


Previously, we compared **means** using *t*-tests. Now we will compare proportions, such as :
  * conversion rates in an A/B test
  * fraud-detection rates of two systems
  * default rates for two groups
  * errors rates of two models

The outcome must be binary: success/failture, fraud/not fraud, converted/not converted, and so on


## Sample proportions


Suppose:
  * Group A has $x_A$ successes among $n_A$ observations
  * Group B has $x_B$ successes among $n_B$ observations

Their sample proprotions are:
$$\hat{p}_A = \frac{x_A}{n_A}, \quad \hat{p}_B = \frac{x_B}{n_B} $$

The observed difference is:
$$ \hat{p}_B - \hat{p}_A $$

The order matters. If we calculate B - A, a positive result means that Group B has the higher rate


## Two-proportion $z$-test


For a two-sided test:
$$ H_0 : p_B - p_A = 0, $$
$$ H_1 : p_B - p_A \neq 0 $$

The null hypothesis says that the population proportions are equal


### Step 1: Calculate the pooled proportion

Under $H_0$, we assume that both groups have one common population proportion. Therefore, we combine their results:

$$ \hat{p}_{pool} = \frac{x_A + x_B}{n_A + n_B} $$


### Step 2: Calculate the standard error under $H_0$

$$ SE_{H_0} = \sqrt{\hat{p}_{pool}(1 - \hat{p}_{pool}) \left( \frac{1}{n_A} + \frac{1}{n_B} \right)} $$


### Step 3: Calculate the test statistic

$$ z = \frac{\hat{p}_B - \hat{p}_A}{SE_{H_0}} $$

For a two-sided test with α = 0.05

$$ z_{critical} = \pm 1.96 $$

Therefore:
  * if $|z| > 1.96$, reject $H_0$
  * if $|z| \leq 1.96$ fail to reject $H_0$


## Example

Suppose:
$$ x_A = 100, \quad n_A = 1000 $$
$$ x_B = 130, \quad n_B = 1000 $$

The sample proportions are:

$$ \hat{p}_A = \frac{100}{1000} = 0.10 $$
$$ \hat{b}_B = \frac{130}{1000} = 0.13 $$

The observed difference is:

$$ \hat{p}_B - \hat{p}_A = 0.13 - 0.10 = 0.03 $$

This is an increase of 3 **percentage points**

The pooled proportion is:

$$ \hat{p}_{pool} = \frac{100 + 130}{1000 + 1000} = \frac{230}{2000} = 0.115 $$

The standard error is:

$$ SE_{H_0} = \sqrt{0.115(1-0.115) \left( \frac{1}{1000} + \frac{1}{1000} \right)} \approx 0.01427 $$

Therefore:

$$ z = \frac{0.13 - 0.10}{0.01427} \approx 2.10 $$

Since:

$$ |2.10| > 1.96 $$

we reject $H_0$ at the 5% signnificance level. There is statistically significant evidence that the two population proportions are different


## Confidence interval for the difference

For a confidence interval, we do **not** pool the proportions:

$$ SE_{CI} = \sqrt{\frac{\hat{p}_A(1 - \hat{p}_A)}{n_A} + \frac{\hat{p}_B(1-\hat{p}_B)}{n_B}} $$

The 95% confidence interval is:

$$ (\hat{p}_B - \hat{p}_A) \pm 1.96SE_{CI} $$

For our example:

$$ SE_{CI} = \sqrt{\frac{0.10(0.90)}{1000} + \frac{0.13(0.87)}{1000}} \approx 0.01425 $$

Thus:

$$ 0.03 \pm 1.96(0.01425) = 0.03 \pm 0.02793 $$

$$ CI_{95\%} \approx [0.0021, 0.0579] $$

The interval does not contain zero, which agrees with the hypothesis-test result

We estimate that Group B's population rate is betweeb approximately **0.21 and 5.79 percentage points higther** than Group A's rate.


## Pooled versus upooled standard errors

| Calculation         | Standard error |
| ---                 | ---            |
| Hypothesis test     | Pooled         |
| Confidence interval | Unpooled       |

We pool in the hypothesis test because $H_0$ assumes:

$$ p_A = p_B $$

We do not pool for the confidence interval because we are estimating the difference without assuming that the proportions are equal


## Percentage points versus relative increase

An increase from 10% to 13% is:

$$ 13% - 10% = 3 $$

percentage points

The ralative increase is:

$$ \frac{13% - 10%}{10%} = 0.30 = 30% $$

Therefore:

> The rate increased by 3 percentage points, or by 30% relative to its original value

These are different statements


## Assumptions

The two-propotion $z$-test requires:
  * a binary outcome
  * two independent groups
  * independent observations within each group
  * sufficiently mant success and failures

A common rule is that each group should contain at least 10 success and 10 failures
