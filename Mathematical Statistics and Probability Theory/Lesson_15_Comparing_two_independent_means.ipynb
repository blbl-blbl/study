# Lesson 15: Comparing two independent means

We already know how to compare one sample with a fixed value:
$$ H_0 : \mu = \mu_0 $$

Now suppose we want to compare two independent groups:
  * group A: old anti-fraud model
  * group B: new anto-fraud model

We test:
$$ H_0 : \mu_A - \mu_B = 0 $$

against, for example:
$$ H_1 : \mu_A - \mu_B \neq 0 $$


## Difference between sample means


The observed difference is:
$$ \bar{X}_A - \bar{X}_B $$

Under H_0, the expected difference is zero

Each sample mean has its own uncertainty:
$$ Var(\bar{X}_A) = \frac{s^2_A}{n_A}, \quad Var(\bar{X}_B) = \frac{s^2_B}{n_B} $$

For independent groups, the standard error of their difference is:
$$ SE(\bar{X}_A - \bar{X}_B) = \sqrt{\frac{s^2_A}{n_A} + \frac{s^2_B}{n_B}} $$

Notice that we add the **variances**, not the standard deviation


## Welch's *t*-test


The test statistic is:
$$ t = \frac{(\bar{X}_A - \bar{X}_B) - 0}{\sqrt{\frac{s^2_A}{n_A} + \frac{s^2_B}{n_B}}} $$

More generally, if $H_0$ assumes a difference $\Delta_0$:
$$ t = \frac{(\bar{X}_A - \bar{X}_B) - \Delta_0}{\sqrt{\frac{s^2_A}{n_A} + \frac{s^2_B}{n_B}}} $$

Usually:
$$ \Delta_0 = 0 $$

This version is called **Welch's *t*-test**. It does not require the two populations to have equal variances, so it's generally safer than the pooled *t*-test

The degrees of freedom are approximately:
$$ df = \frac{\left( \frac{s^2_A}{n_A} + \frac{s^2_B}{n_B} \right)^2}{ \frac{(s^2_A/n_A)^2}{n_A-1} + \frac{(s^2_B / n_B)^2}{n_B-1}} $$

In practice, software normally calculates this value


## Example


Suppose transaction-review times are:
$$ Old~model:~n_A = 50, \bar{X}_A = 12, s_A = 4, $$
$$ New~model:~n_B = 50, \bar{X}_B = 10, s_B = 5 $$

We test:
$$ H_0: \mu_A - \mu_B = 0, $$
$$ H_1: \mu_A - \mu_B \neq 0 $$

The observed error is:
$$ SE = \sqrt{\frac{4^2}{50} + \frac{5^2}{50}} = \sqrt{0.32 + 0.50} = \sqrt{0.82} \approx 0.906 $$

Therefore:
$$ t_{critical} = \pm 1.986, $$

Because:
$$ |2.21| > 1.986, $$

we reject $H_0$

There is statistically signficant evidence that the population means differ


## Confidence interval for the difference


The confidence intarval is:
$$ (\bar{X}_A - \bar{X}_B) \pm t_{critical} \cdot SE $$

For our example:
$$ 2 \pm 1.986 \cdot 0.906 = 2 \pm 1.799 = [0.201, 3.799] $$

The interval does not contain zero, so it gives the same conclusion as the hypothesis test

The estimated difference is positive, meaning that the old model requires approximately 0.2 to 3.8 more minutes on average


## Assumptions


Welch's test assumes:
  * the two groups are independent
  * observations within each group are independent
  * the outcome is quantative
  * the sampling distribution of each mean is approximately normal

For small samples, severe skewness or large outliers may cause problems. Welch's test does not require equal population variances


## Independent versus paired observations


The independent test is appropriate when the groups contain different objects or people:

Example:
  * different customers in groups A and B
  * different transactions processed by two models
  * separate control and tratment groups

It is not appropriate when the observations are naturally matched:
  * the same customers before and after a change
  * the same companies measured in two years
  * predictions from two models on exactly the same transactions

Matched observations require a **paired *t*-test**, which we will study next

### Task 1

Two independent groups have:

$$ n_A = 64, \quad \bar{X}_A = 100, \quad s_A = 16, $$
$$ n_B = 49, \quad \bar{X}_B = 94, \quad s_B = 14 $$

Test:
$$ H_0 : \mu_A - \mu_B = 0 $$
$$ H_1 : \mu_A - \mu_B \neq 0 $$

at α = 0.05

Use:
$$ t_{critical} = \pm 1.98 $$

Calculate:
  1. the observed difference
  2. the standard error of the difference
  3. the *t*-statistic
  4. the statistical decision
  5. the 95% confidence interval for $\mu_A - \mu_B$


**Solution**:

  1. the observed difference is:
  $$ 100 - 94 = 6 $$

  2. the standard error of the difference:
  $$ SE(\bar{X}_A - \bar{X}_B) = \sqrt{\frac{16^2}{64} + \frac{14^2}{49}} = \sqrt{4 + 4} \approx 2.83  $$

  3. the *t*-statistic:
  $$ t = \frac{6}{2.83} = 2.12 $$

  4. the statistical decision:
  
  Because:
  $$ |2.12| > 1.98 $$
  we reject $H_0$

  5. the 95% confidence interval for $\mu_A - \mu_B$
  $$ CI_{95\%} = 6 \pm 1.96 \cdot 2.83 = [6 - 5.54, 6 + 5.54] = [0.56, 11.54] $$

  The intarval does not contain zero, so it gives the same conclusion as the hypothesis test
