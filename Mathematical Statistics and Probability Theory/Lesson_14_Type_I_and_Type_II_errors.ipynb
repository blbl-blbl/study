# Lesson 14: Type I and Type II errors

A hypothesis test can produce two types of errors

| Reality | Reject $H_0$ | Fail to reject $H_0$ |
| :--- | :--- | :--- |
| $H_0$ is true | Type I error | Correct decision |
| $H_0$ is false | Correct decision | Type II error |

## Type I error

A type I error occurs when we reject a true null hypothesis - a false positive

Its probability:
$$ P(Type~I~error) = \alpha $$

For example:
$$ H_{0}:~a~new~anti-fraud~rule~does~not~improve~detection $$

A Type I error means concluding that the rule improves detection when it actually does not


## Type II error

A Type II error occurs when we fail to reject a false null hypothesis - a false negative

Its probability is:
$$ P(Type~II~error) = \beta $$

In the previous example, this means failing to detect that the new rule geneuinely improves fraud detection


## Statistical power

Power is the probability of detecting an effect when it really exists:

$$ Power = 1 - \beta $$

If:

$$ \beta = 0.20, $$

then:

$$ Power = 0.80 $$

The test has an 80% probability of detecting the specified effect

Power increases when:
  * the sample size increases
  * the true effect becomes larger
  * variability decreases
  * α increases
  * an appropriate one-sided test is used

Reducing α decreases the risk of a Type I error, but - if everything else remains fixed - it increases the risk of a Type II error

Increasing the sample size helps reduce both risks without increasing α

## Statistical versus practical significance

A very small effect can become statistically significant with a sufficiently large sample

For example:
$$ old~conversion~rate = 10.00\% $$
$$ new~conversion~rate = 10.01\% $$

With millions of observations, the difference might be statistically significant but practically unimportant

Therefore, we should examine both:
  * the p-value
  * the magnitude of the effect and its confidence interval

A bank tests a new anti-fraud model:

$$ H_{0}: the~new~model~does~not~improve~fraud~detection $$
$$ H_{1} the~new~model~improves~fraud~detection $$

Explain what the following mean:
  1. a Type I error
  2. a Type II error
  3. a correct rejection of $H_0$
  4. a correct failure of reject $H_0$


**Answers:**
1. In this case Type I error means that statistical tests show that the new model does improve fraud detection, but in reality it does not
2. When statistical test show that model does not improve fraud detection, but in reality it does
3. If model improves fraud detection, we can reject $H_0$
4. If model doesn't improve fraud detection - its a correct falure of reject $H_0$

## Task 3

A right-sided Z-test has:
$$ H_{0} : \mu = 50, \quad H_{1} : \mu > 50, $$

$$ \sigma = 10, \quad n = 25, \quad \alpha = 0.05 $$

Use:
$$ z_{critical} = 1.645 $$

1. Find the critical sample mean above which $H_0$ is rejected:

$$ \bar{X}_{critical} = \mu_0 + z_{critical} \frac{\sigma}{\sqrt{n}} $$

2. Suppose the true mean is actually μ = 55. Calculate the probability of rejecting $H_0$. Use:

$$ Φ(0.855) \approx 0.803 $$

This probability is the power of the test against the specific alternative μ = 55


**Solution**

1. Critical sample mean above which $H_0$ is rejected:

$$ \bar{X}_{critical} = 50 + 1.645 \frac{10}{\sqrt{25}} = 50 + 3.29 = 53.29 $$

2. Calculating the probability of rejecting $H_0$ with μ = 55

Based on previous point we know that the test decision rule is:

$$ Reject~H_0~if~\bar{X} > 53.29 $$

$$ Z = \frac{53.29-55}{10 / \sqrt{25}} = -0.855 $$

$$Power = \Phi(0.855) = 0.803 = 80.3\% $$

$$ \beta = 1 - 0.803 = 0.197 = 19.7\% $$
