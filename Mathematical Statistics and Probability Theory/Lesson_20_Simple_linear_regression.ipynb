# Lesson 20: Simple Linear Regression

Correlation describes the strength and direction of a linear relationship. Linear regression creates an equation that can be used for explanation and prediction

The simple linear regression model is:

$$ Y = \beta_0 + \beta_1X + \epsilon, $$

where:

  * $\beta_0$ is the population intercept
  * $\beta_1$ is the population slope
  * $\epsilon$ is random error

From a sample, we estimate:

$$ \hat{Y} = b_0 + b_1X $$


## Slope

The estimated slope is:

$$ b_1 = \frac{\sum(x_i - \bar{x})(y_i-\bar{y})}{\sum(x_i - \bar{x})^2} $$

It cal also be written as:

$$ b_1 = \frac{Cov(X, Y)}{s_X^2} $$

The slope represents the expected change in *Y* when *X* increases by one unit


## Intercept

The estimated intercept is:

$$ b_0 = \bar{y} - b_1\bar{x} $$

It represents the predicted value of *Y* when:

$$ X = 0$$

Sometimes $X = 0$ is not meaningful or is outside the observed data, so the intercept may not always have a useful practical interpretation

## Example

Using the previous data:

$$ X = [1, 2, 3, 4, 5] $$

$$ Y = [50, 55, 65, 70, 80] $$

We already know:

$$ \bar{x} = 3, \quad \bar{y} = 64 $$

We also calculated:

$$ \sum(x_i - \bar{x})(y_i - \bar{y}) = 75 $$

$$ \sum(x_i - \bar{x})^2 = 10 $$

Therefore:

$$ b_1 = \frac{75}{10} = 7.5 $$

The intercept is:

$$ b_0 = 64 - 7.5 \cdot 3 = 41.5 $$

The regression equation is:

$$ \hat{Y} = 41.5 + 7.5X $$

Interpretation:

  `For every one-unit increase in X, the predicted value of Y increases by 7.5 units on average`

For $X = 4$:

$$ \hat{X} = 41.5 + 7.5 \cdot 4 = 71.5 $$


## Residuals

A residual is the difference between the observed and predicted values

$$ e_i = y_i - \bar{y}_i $$

For $X = 4$:

$$ e = 70 - 71.5 = -1.5 $$

The model overestimated the observed value by 1.5


## Sum of squared errors

The sum of squared residuals is:

$$ SSE = \sum(y_i - \hat{y})^2 $$

The least-squares regression line chooses $b_0$ and $b_1$ so that the SSE is as small as possible


## Coefficient of determination

The coefficinet of determination is:

$$ R^2 = 1 - \frac{SSE}{SST}, $$

where:

$$ SST = \sum(y_i - \bar{y})^2 $$

For the example:

$$ SST = 570 $$

$$ R^2 = 1 - \frac{7.5}{570} \approx 0.9868 $$

Interpretation:

  `Approximately 98.68% of the variation in Y is explained by its linear relationship with X in this sample`

For simple linear regreesion with an intercept:

$$ R^2 = r^2 $$

Here:

$$ 0.9934^2 \approx 0.9868 $$


## Extrapolation

The model predicts:

$$ \hat{Y} = 41.5 + 7.5 \cdot 6 = 86.5 $$

for $X = 6$

However, the observed range is only:

$$ 1 \leq X \leq 5 $$

Predicting outside the observed range is called **extrapolation**. It is risker because the relationship may change outside the available data


## Main assumptions

Simple linear regression assumes:

  * a linear relationship
  * independent observations
  * approximately constant residual variance
  * no extreme influential outliers
  * approximately normal residuals when performing hypothesis tests and confidence intervals

### Task 2. Conceptual questions

1. What does a negative slope mean?

  - It means that for every one unit increases in X, the predicted value of Y decreases by some value an average. If there are only two variables: X and Y, it also means that there is negative correlation between those variables

2. Does a high $R^2$ prove causation?

  - No, it doesn't It only shows how linear regression can decribes variation in *Y* in relationship with X in this sample

3. If *r* = -0.8, what is $R^2$, and will the slope be positive or negative?

  - Slope will be negative. Meaning of $R^2$ isn't changing in this case, it will describe how the linear regression can describe variation in *Y*


4. Why can the intercept sometimes have no practical meaning?

  - Because this is linear model. For example, in some cases it even can be negative, but negative value in this cases will be meaningless and just can't be real

5. Why is extrapolation risky?

  - Because this is prediction on some data which we can't really prove

6. What patterns in a residual plot may indicate that linear regression is unsuitable?

  - If we can see some linear trends on plot - we can conclude, that linear regression missed relationships between variables. If we see that there are a lot of "big" residuals, we can assume that there is a non-linear relationship between varibales and linear regression is unsuitable. For an appropriate linear model, residuals should look like a random cloud around zero
