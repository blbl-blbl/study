# Lesson 25: ROC-AUC, PR-AUC, and threshold selection

A classifier normally produces a score or probability:

$$ \hat{p} = P(Y = 1 | X) $$

We convert it into a class using a threshold *t*:

$$ \hat{Y} = \begin{cases} 1, \ \hat{p} \geq t, \\ 0, \ \hat{p} < t \end{cases} $$

Changing *t* changes the confusion matrix


## ROC curve

For every threshold, calculate:

$$ TPR = \frac{TP}{TP + FN} = Recall $$

and

$$ FPR = \frac{FP}{FP + TN} $$

The ROC curve plots:

  - $FPR$ on the horizontal axis
  - $TPR$ on the vertical axis

A useful threshold has high TPR and low FPR


## ROC-AUC

ROC-AUC summarizes the entire ROC curve:

$$ 0 \leq AUC \leq 1 $$

Typical interpretation:

  - $AUC = 1$: perfect ranking
  - $AUC = 0.5$: approximately random ranking
  - $AUC < 0$: ranking is worse than random

AUC evalulates ranking across thresholds. It does not tell us which threshold to use


### Pairwise interpretation

Suppose fraud scores are:

$$ 0.9, \quad 0.7 $$

and legitimate-transaction scores are:

$$ 0.8, \quad 0.4, \quad 0.2 $$

There are:

$$ 2 \cdot 3 = 6 $$

fraud-legitmate pairs

The score 0.9 exceeds all three legitmate scores. The score 0.7 exceeds 0.4 and 0.2, but not 0.8. Therefore, the model ranks five of the six pairs correctly:

$$ AUC = \frac{5}{6} \approx 0.8333 $$



## Precision-recall curve

The precision-recall curve plots:

  - recall on the horizontal axis
  - precison on the vertical axis

PR-AUC can be especially useful when the positive class is rare. Its random-model baseline is approximately the positive-class prevalence

If fraud prevalence is 2%, then:

$$ PR\text{-}AUC_{random} \approx 0.02 $$

Consequently, a PR-AUC of 0.18 would be much better then random, even though 0.18 may initially appear small



## Youden's $J$

One possible threshold-selection metric is:

$$ J = TPR - FPR $$

However, Youden's $J$ treats sensitivity and specificity as equally important. That may not match the company's actual costs


## Cost-based threshold selection

When business costs are available, we can select the threshold that minimizes:

$$ Cost = FN \cdot C_{FN} + FP \cdot C_{FP} $$

If predicted probabilities are well calibrated and these are the only two costs, the theoretical decision threshold is:

$$ t = \frac{C_{FP}}{C_{FP} + C_{FN}} $$

For:

$$ C_{FN} = 300, \quad C_{FP} = 10 $$

we obtain:

$$ t = \frac{10}{10 + 300} \approx 0.0323 $$

This low threshold is reasonable because missing fraud is much more expensicve than investigating a false alert

In practice, the final threshold should be tested on validation data because probabilities may not be perfectly calibated and real business costs may be more complicated

### Task 1

Three thresholds produce the following results:

| Result | T1 | T2 | T3 |
| :--- | :--- | :--- | :--- |
| TP | 40 | 70 | 90 |
| FN | 60 | 30 | 10 |
| FP | 10 | 60 | 210 |
| TN | 890 | 840 | 690 |

For every threshold, calculate:

  1. recall
  2. FPR
  3. precision
  4. Youden's $J$
  5. which threshold has the largest $J$


**Solution:**

1. Recall (TPR)

$$ Recall_{T1} = \frac{40}{40 + 60} = 0.4 $$

$$ Recall_{T2} = \frac{70}{70 + 30} = 0.7 $$

$$ Recall_{T3} = \frac{90}{90 + 10} = 0.9 $$


2. FPR

$$ FPR_{T1} = \frac{10}{10 + 890} \approx 0.011 $$

$$ FPR_{T2} = \frac{60}{60 + 840} \approx 0.0667 $$

$$ FPR_{T3} = \frac{210}{210 + 690} \approx 0.233 $$


3. Precision

$$ Precision_{T1} = \frac{40}{40 + 10} = 0.8 $$

$$ Precision_{T2} = \frac{70}{70 + 60} \approx 0.538 $$

$$ Precision_{T3} = \frac{90}{90 + 210} = 0.3 $$


4. $J$

$$ J_{T1} = 0.4 - 0.011 \approx 0.389 $$

$$ J_{T2} = 0.7 - 0.0667 \approx 0.633 $$

$$ J_{T3} = 0.9 - 0.233 \approx 0.667 $$


5. third threshold has the largest $J$



### Task 2

Assume:

Using the confusion matrices from Task 1:

  1. calculate the business cost of each threshold
  2. select the least expensive threshold
  3. explain why it may differ from the threshold with the highest accuracy
  4. calculate the theoretical probability threshold


**Solution:**

1. Business cost

$$ Cost_{T1} = 200 \cdot 60 + 10 \cdot 10 = 12 \ 100 $$

$$ Cost_{T2} = 200 \cdot 30 + 10 \cdot 60 = 6 \ 600 $$

$$ Cost_{T3} = 200 \cdot 10 + 10 \cdot 210 = 4 \ 100 $$

2. third threshold is the least expensive

3. because cost of missing positive signal is more expensive than false positive, so that is the reason why recall is more important than precision

$$ t = \frac{10}{10 + 200} \approx 0.0476 $$

### Task 4

1. Does a high ROC-AUC automatically give us the best classification threshold?

  - No, it doesn't, to select a threshold we should look not only on classification metrics, but at the business metrics too, and the last metrics usually is more important in threshold selecting

2. Why can ROC-AUC appear strong when fraud is extremely rare?

  - Because ROC-AUC is a share of correctly classification pairs, but in inbalanced data it can be a problem, because a larger share would be correctly determined

3. What is the approximate random PR-AUC baseline when prevalence is 1%?

  - Around 0.01

4. Can two models have similar ROC-AUC values but very different business costs?

  - Yes, it is poosible

5. What normally happens to precision and reacll when the threshold is lowered?

  - When threshold is lowered usually recall increasing and precision decreasing
