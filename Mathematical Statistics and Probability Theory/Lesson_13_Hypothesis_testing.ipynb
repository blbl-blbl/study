# Lesson 13: Hypothesis testing

A statistical hypothesis test evaluates whether sample evidence is inconsistent with a claim about a population

## Null and alternative hypotheses

The **null hypothesis** represents the initial claim:

$$ H_{0}: \mu = \mu_{0} $$

The **alternative hypothesis** represents what we want to investigate

A two-sided alternative hypothesis is:
$$ H_{1}: \mu \neq \mu_{0} $$

One-sided alternatives are:
$$ H_{1}: \mu > \mu_{0} $$
or:
$$ H_{1} : \mu < \mu_{0} $$

The direction must be selected before examining the results


## Test statistic

If σ is known:
$$ Z = \frac{\bar{X} - \mu_{0}}{\sigma / \sqrt{n}} $$

If σ is unknown:
$$ T = \frac{\bar{X} - \mu_{0}}{s / \sqrt{n}}, \quad df = n - 1 $$

The statistic measures how many standard errors the observed sample mean is from the value assumed under H<sub>0</sub>


## The p-value

The p-value is:

  | The probability, assuming H<sub>0</sub> is true, of obtaining a result at least as extreme as the observed result

It is not the probability that H<sub>0</sub> is true.

We choose a sagnificance level, commonly:
$$ \alpha = 0.05 $$

The decision rule is:
  * if p ≤ α, reject H<sub>0</sub>
  * if p > α, fail to reject H<sub>0</sub>

"Fail to reject" does not mean that H<sub>0</sub> has been proven true. It means that the evidence against it is insufficient


## Example

Suppose:

$$ \bar{X}, \quad \mu_{0}=100, \quad s = 15, \quad n = 25 $$

We test:

$$ H_{0} : \mu = 100, \quad H_{1} : \mu \neq 100 $$

Then:

$$ T = \frac{106 - 100}{15 / \sqrt{25}} = \frac{6}{3} = 2 $$

$$ df = 24 $$

For the two-sided test with α = 0.05:
$$ t_{critical} = \pm 2.064 $$

Because:
$$ |2| < 2.064, $$
we fail to reject H<sub>0</sub>

The corresponding confidence interval is:

$$ 106 \pm 2.064 \cdot 3 = [99.808, 112.192] $$

It contains 100, giving the same conclusion

### Task 1

Suppose:

$$ \mu_{0} = 50, \quad \bar{X}=53, \quad \sigma=15 \quad n=100 $$

Test:
$$ H_{0}: \mu = 50, \quad H_{1}: \mu > 50 $$

at:
$$ \alpha = 0.05 $$

Calculate:
  1. the standard error
  2. the Z-statistic
  3. the p-value, using 1 - Φ(2) = 0.0228
  4. the conclusion


Solution:

1. The standard error:
$$ SE = \frac{15}{\sqrt{100}} = 1.5 $$

2. The Z-statistic:
$$ Z = \frac{53 - 50}{15 / \sqrt{100}} = \frac{3}{1.5} = 2 $$

3. The p-value
$$ pvalue = 1 - Φ(2) = 0.0228 $$

4. The conclusion:
For right-sided:
$$ Z_{crit} = 1.65 $$

Because
$$ 2 > 1.65 \quad ⇔ \quad p_{value} < α $$

we reject H<sub>0</sub>
