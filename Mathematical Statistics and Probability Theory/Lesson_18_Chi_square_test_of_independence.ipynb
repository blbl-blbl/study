# Lesson 18: Chi-square test of independence

The chi-square test examines thether two categorical varibles are associated

Examples:
  * device type and purchase decision
  * risk group and default
  * customer segment and product choice
  * model type and correct/incorrect prediction


## Contingency table

Suppose we observe:
| System | Detected   | Missed | Total |
| ---    | ---        | ---    | ---   |
| Old    | 180        | 820    | 1000  |
| New    | 215        | 785    | 1000  |
| Total  | 395        | 1605   | 2000  |

These are called **observed frequencies**, written as $O$


## Hypotheses

$$ H_0 : The~two~categorical~variables~are~independent $$
$$ H_1 : The~two~categorical~variables~are~associated $$

In this example:

$$ H_0 : Detection~result~is~independent~of~system~type $$


## Expected frequencies

If the variables are independent, the expected frequence in each cell is:

$$ E_{ij} = \frac{(row~total)(column~total)}{grand~total} $$

For the old system and detected transactions:

$$ E = \frac{1000 \cdot 395}{2000} = 197.5 $$


The complete expected table is:
| System | Detected | Missed |
| --- | --- | --- |
| Old | 197.5 | 802.5 |
| New | 197.5 | 802.5 |


## Chi-square statistic

$$ \chi^2 = \sum \frac{(O-E)^2}{E} $$

For this table:

$$ \chi^2 = \frac{(180-197.5)^2}{197.5} + \frac{(820-802.5)^2}{802.5} + \frac{(215-197.5)^2}{197.5} + \frac{(785-802.5)^2}{802.5} \approx 3.865 $$

The chi-square statistic is always non-negative. Larger values mean that the observed table is further from what independence predicts


## Degrees of freedom

For a table with $r$ rows and $c$ columns:

$$ df = (r-1)(c-1) $$

For a 2 × 2 table:

$$ df = (2-1)(2-1) = 1 $$

At α = 0.05 and $df$ = 1:

$$ \chi^2_{critical} = 3.841 $$

Since:

$$ 3.865 > 3.841 $$

we reject $H_0$. System type and detection result are statistically assosiated

Notice:

$$ \chi^2 = z^2 \approx 1.966^2 \approx 3.865 $$

For a 2 × 2 table, the chi-square test and the two-sided two-proportion $z$-test produce equivalent results.


## Effect size: Cramer's V

Statistical significamce does not show whether the association is practically important

Cramer's *V* is:

$$ V = \sqrt{\frac{\chi^2}{n~min(r-1, c-1)}} $$

It ranges from 0 to 1:
  * 0: no association
  * values closer to 1: stronger association

For our example:

$$ V = \sqrt{\frac{3.85}{2000 \cdot 1}} \approx 0.044 $$

The association is statistically significant but small


## Assumptions

The chi-square test requires:
  * categorical data represented as counts
  * independent observations
  * mutually exclusive categories
  * sufficiently large expected frequencies

A common rule is:
$$ E_{ij} \geq 5 $$

for every cell. For small expected frequencies, Fisher's exact test may be preferable

### Task 1

A company studies wheter purchasing is associated with device type:

| Device  | Purchase | No purchase | Total |
| ---     | ---      | ---         | ---   |
| Mobile  | 120      | 280         | 400   |
| Desktop | 180      | 220         | 400   |
| Total   | 300      | 500         | 800   |

Calculate:
  1. the row totals, column totals and grand total
  2. all expected frequencies
  3. the hypotheses
  4. the chi-square statistic
  5. the degrees of dreedom
  6. the decision using $\chi^2_{critical} = 3.841$
  7. the purchase rate for each device
  8. Cramer's $V$
  9. whether the test alone proves type causes the difference


**Solution:**

1. Totals added to table
2. all expected frequencies

| Device  | Purchase | No purchase |
| ---     | ---      | ---         |
| Mobile  | 150      | 250         |
| Desktop | 150      | 250         |

3. the hypotheses

$$ H_0 : The~two~categorical~variables~are~independent $$

$$ H_1 : The~two~categorial~variables~are~associated $$

4. the cho-square

$$ \chi^2 = \frac{(120 - 150)^2}{150} + \frac{(280 - 250)^2}{250} + \frac{(180 - 150)^2}{150} + \frac{(220 - 250)^2}{250} = 19.2  $$

5. the degrees of freedom

$$ df = (2-1)(2-1) = 1 $$

6. Since:
$$ 19.2 > 3.841 $$

we reject $H_0$. Device and fact of purchase are statistically associated

7.
$$ Mobile~purchase~rate = \frac{120}{400} = 0.3 = 30\% $$

$$ Desktop~purchase~rate = \frac{180}{400} = 0.45 = 45\% $$

8. Cramer's $V$

$$ V = \sqrt{\frac{19.2}{800 \cdot 1}} \approx 0.1549 $$

9. The association is statistically significant but not very high
