# Lesson 12: Student's t-distribution

Usually, the pipulation standard deviation σ is unknown. We replace it with the sample standard deviation s:

$$ \bar{X} \pm t_{\alpha/2,n-1} \frac{s}{\sqrt{n}} $$

The number:

$$ df = n - 1 $$

is called the **degrees of freedom**


The t-distribution has heavier tails than the normal distribution because using `s` introduces additional uncertainly. Therefore, its critical value is larger:

| Degrees of freedom | 95% t-critical value |
| :---: | :---: |
| 4 | 2.776 |
| 9 | 2.262 |
| 19 | 2.093 |
| 29 | 2.045 |
| ∞ | 1.960 |


As the sample size increases, the t-distribution approaches the standard normal distribution

### Example

Suppose:

$$ \bar{X} = 52, \quad s = 20, \quad n =25 $$

Then:

$$ df = 24, \quad t_{0.025, 24} = 2.064 $$

$$ SE = \frac{20}{\sqrt{25}} = 4 $$

Therefore:

$$ CI_{95\%} = 52 \pm 2.064 \cdot 4 = 52 \pm 8.256 = [43.744, 60.256] $$

## Practice

### Task 1

Suppose:

$$ \bar{X} = 80, \quad s=12, \quad n = 16 $$

Use:

$$ t_{0.025, 15} = 2.131 $$

Calculate:

1. the degrees of freedom;
2. the standard error;
3. the margin of error;
4. the 95% confidence interval.


Decision:

1. The degrees of freedom:
$$ df = 16 - 1 = 15 $$

2. The standard error:
$$ SE = \frac{12}{\sqrt{16}} = 3 $$

3. The margin of error:

$$ ME = 3 \cdot 2.131 = 6.602 $$

4. The 95% confidence interval:

$$ CI_{95\%} = 80 \pm 6.602 = [73.398, 86.602] $$
