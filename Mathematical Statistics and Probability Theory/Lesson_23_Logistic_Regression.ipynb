# Lesson 23: Logistic Regression

Linear reggresion predicts a numerical outcome. Logistic regression predicts the probability of a binary outcome:

$$ Y \in \{0, 1\} $$

Examples include:

  * fraud or not fraud
  * default or no default
  * customer converts or does not convert
  * desease present or absent


## Why not use linear regression?

A linear model could predict:

$$ \hat{p} = -0.2 $$

or

$$ \hat{p} = 0.13 $$

These cannot be probabilities. Linear regression also does not represent the typical curved relationship between predictors and probabilities

Logistic regression solves this using the sigmoid function


## Sugmoid function

First, the model calculates a linear value:

$$ z = \beta_0 + \beta_1X_1 + ... + \beta_kX_k $$

It then transforms $z$ into a probability:

$$ p = \frac{1}{1 + e^{-z}} $$

The result always satifies:

$$ 0 < p < 1 $$

If:

$$ z = 0 $$

then:

$$ p = \frac{1}{1 + e^{0}} = 0.5 $$

Positive $z$ produces $p > 0.5$, and negative $z$ produces $p < 0.5$



## Odds

Probability and odds are related but different

$$ Odds = \frac{p}{1 - p} $$

If:

$$ p = 0.2 $$

then:

$$ Odds = \frac{0.2}{1 - 0.2} = 0.25 $$

This can be described as odds of 1 to 4

We can recover probability from odds:

$$ p = \frac{Odds}{1 + Odds} $$



## Log-odds

Logistic regression models the logarithm of the odds:

$$ log \left( \frac{p}{1 - p} \right) = \beta_0 + \beta_1 X_1 + ... + \beta_k X_k $$

The left side is called the **logit** or **log-odds**



## Coefficient interpretation

A one-unit increase in $X_j$ changes log-odds by $\beta_j$, holding other predictors constant

Usually, we interpret:

$$ e^{\beta_j}, $$

which is the **odds ratio**

Suppose:

$$ \beta_j = 0.8 $$

Then:

$$ e^{0.8} \approx 2.23 $$

Interpretation:

*Holding other predictors constant, a one-unit increase in $X_j$, multiplies the odds of the event by approximately 2.23*

This is a 123% increase in odds:

$$ (2.23 - 1) \cdot 100\% = 123\% $$

If:

$$ \beta_j = -0.5 $$

then:

$$ e^{-0.5} \approx 0.607 $$

The odds are multiplied by 0.607, meaning they decrease by approximately:

$$ (1 - 0.607) \cdot 100\% = 39.3\% $$

A logistic coefficient does not represent a constant change in probability. Its effect on probability depends on the starting probability



## Example

Suppose the fraud model is:

$$ log \left( \frac{p}{1 - p} \right) = -4 + 0.7 \text{Amount} - 0.5 \text{Age} $$

where:
  * *Amount* is transaction amount in tens of thousands
  * *Age* is accaunt age in years

For:

$$ Amount = 5, \quad Age = 2 $$

we first calculate:

$$ z = -4 + 0.7(5) - 0.5(2) = -1.5 $$

Then:

$$ p = \frac{1}{1 + e^{1.5}} \approx 0.1824 $$

The model predicts an 18.24\% probability of fraud

The odds are:

$$ Odds = e^{-1.5} \approx 0.2231 $$



## Classification threshold

A common rule is:

$$ \hat{Y} = \begin{cases} 1,&  \geq 0.5 \\ 0,& p <0.5 \end{cases}$$

Because:

$$ 0.1824 < 0.5 $$

the transaction would be classified as non-fraudlent

Hower, 0.5 is not always the best threshold. In fraud detection, missing fraud may be much more expensive than investigating a false alert, so a lower threshold may be appropriate



## Testing coefficients

For each predictor:

$$ H_0 : \beta_j = 0 $$

$$ H_1 : \beta_j \neq 0 $$

A common test statistic is:

$$ z = \frac{b_j}{SE(b_j)} $$

At $\alpha = 0.05$, we often compare with:

$$ \pm 1.96 $$

A confidence interval for a coefficient is:

$$ b_j \pm 1.96 SE(b_j) $$

Exponentiating both limits gives a confidence interval for the odds ratio
